```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("bhanupratapbiswas/customer-behavior-analysis")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os

files = os.listdir(path)   # gives list
print("Folder", files)

csv_path = os.path.join(path, files[0])
df = pd.read_csv(csv_path)

print(df.info())
```

```python
df['Returns'] = df['Returns'].fillna(0)
print("Both columns are same?", df['Customer Age'].equals(df['Age']))
```

```python
df['Purchase Date'] = pd.to_datetime(df['Purchase Date'])
print(df['Purchase Date'].dtype)
```

```python
df = df.drop(columns = ['Age'], errors = 'ignore')
df['Purchase_Month'] = df['Purchase Date'].dt.month
print(df[['Purchase Date', 'Purchase_Month']].head())
```

```python
monthly_category_sales = df.groupby(['Purchase_Month', 'Product Category'])['Total Purchase Amount'].sum().reset_index()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(18, 9))
ax = sns.barplot(
     data = monthly_category_sales,
     x = 'Purchase_Month',
     y = 'Total Purchase Amount',
     hue = 'Product Category'
)



plt.title("Trends of Monthly Sales by Product Category (Purchase Patterns)", fontsize = 14, fontweight = 'bold' )
plt.xlabel("Months (1-12)", fontsize = 12)
plt.ylabel("Total Revenue", fontsize = 12)
plt.grid(True, linestyle= '--', alpha = 0.2)

plt.xticks(range(1, 13))

plt.legend(title = 'Product Category', bbox_to_anchor = (1.02, 1), loc = 'upper left')

plt.tight_layout()
plt.show()
```

```python
pivot_sales = df.pivot_table(
    index = 'Purchase_Month',
    columns = 'Product Category',
    values = 'Total Purchase Amount',
    aggfunc = 'sum'
) / 1000000

print (" Monthly sales by Product category:")
print(pivot_sales.round(2))
```

```python
rfm_basic = df.groupby('Customer ID').agg({
    'Total Purchase Amount' : 'sum',
    'Product Category' : 'count'
}).reset_index()

rfm_basic.columns = ['Customer ID', 'Monetary_Value', 'Frequency_Value']

print(rfm_basic.head())
```

```python
def assign_segment(row):
  if row['Frequency_Value'] > 5:
    return 'VIP Customer'
  elif row['Frequency_Value'] >= 3:
    return 'Regular Customer'
  else:
    return 'Low-Value / Risk Customer'

rfm_basic['Customer_Segment'] = rfm_basic.apply(assign_segment, axis = 1)

print(rfm_basic['Customer_Segment'].value_counts())
```

```python
segment_counts = rfm_basic['Customer_Segment'].value_counts()
total_customers = segment_counts.sum()

def make_autopct(values):
  def my_autopct(pct):
    val = int(round(pct * total_customers / 100.0))
    return f'{pct:.1f}%\n({val:,})'
  return my_autopct

plt.figure(figsize=(6,6))

Colors = ['#4CAF50', '#2196F3', '#FF9800']

plt.pie(
    segment_counts,
    labels= segment_counts.index,
    autopct = make_autopct(segment_counts),
    startangle = 140,
    colors = Colors,
    textprops = {'fontsize': 12, 'fontweight': 'bold'},
    pctdistance = 0.75

)

plt.title('Customer Segments Contibution', fontsize = 14, fontweight = 'bold')
plt.tight_layout()
plt.show()
```

```python
category_returns = df.groupby('Product Category')['Returns'].sum().reset_index()

category_returns = category_returns.sort_values(by = 'Returns', ascending = False)

print (" Returns on which Product Category --")
print(category_returns)
```

```python
monthly_return = df.groupby("Purchase_Month")['Returns'].sum().reset_index()

print("Total returns on which months?")
print(monthly_return)
```

```python
monthly_returns = df.groupby(["Purchase_Month", "Product Category"])['Returns'].sum().reset_index()

print("Total returns on which months?")
print(monthly_returns)
```

```python
plt.figure(figsize=(18, 9))
ax = sns.barplot(
     data = monthly_returns,
     x = 'Purchase_Month',
     y = 'Returns',
     hue = 'Product Category'
)



plt.title("How many products category retuns every month? ", fontsize = 14, fontweight = 'bold' )
plt.xlabel("Months (1-12)", fontsize = 12)
plt.ylabel("Total Return", fontsize = 12)
plt.grid(True, linestyle= '--', alpha = 0.2)

plt.xticks(range(1, 13))

plt.legend(title = 'Product Category', bbox_to_anchor = (1.02, 1), loc = 'upper left')

plt.tight_layout()
plt.show()
```
